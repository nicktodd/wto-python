## Working with Merging of Dataframes

````python
# load in a new dataframe from excel
countries = pd.read_excel(
    "data/countries.xlsx")
# merge the trade data with the country income group information
merged = trade_dataframe[["reporter_iso3", "region", "exports_usd_m"]].merge(
    countries[["iso3", "income_group"]],
    left_on="reporter_iso3",
    right_on="iso3", how="left")

print(merged.head(3))

merged.groupby("reporter_iso3")[
    "exports_usd_m"].sum()
````

### What does `how="left"` mean?

`how=` is the join type, exactly as in SQL. The DataFrame you call `.merge()` on is the **left** table; the one you pass in is the **right** table. So the merge above is:

```sql
SELECT t.*, c.income_group
FROM   trade t
LEFT JOIN countries c ON t.reporter_iso3 = c.iso3
```

| `how=`    | SQL                 | Rows kept                                                 |
|-----------|---------------------|-----------------------------------------------------------|
| `"left"`  | `LEFT JOIN`         | Every left row; right columns are NaN where there is no match |
| `"right"` | `RIGHT JOIN`        | Every right row; left columns are NaN where there is no match |
| `"inner"` | `INNER JOIN`        | Only keys present on **both** sides. **This is the default.** |
| `"outer"` | `FULL OUTER JOIN`   | Everything from both sides, NaN where unmatched           |
| `"cross"` | `CROSS JOIN`        | Every left row paired with every right row                |

Two differences from SQL to watch for:

* The default is **inner**, not left. Forget `how=` and unmatched rows silently vanish.
* There is no `WHERE` afterwards. Use `indicator=True` to get a `_merge` column you can filter on.


### A tiny example

Two small tables so every join fits on screen. Ethiopia (`ETH`) has sales but is missing from the lookup; South Africa (`ZAF`) is in the lookup but has no sales.

````python
sales = pd.DataFrame({
    "iso3": ["KEN", "KEN", "ETH"],
    "exports": [10, 20, 5],
})
lookup = pd.DataFrame({
    "iso3": ["KEN", "ZAF"],
    "income_group": ["Lower middle", "Upper middle"],
})

for how in ["left", "right", "inner", "outer"]:
    print(f"--- how='{how}' ---")
    print(sales.merge(lookup, on="iso3", how=how), end="\n\n")
````

### The same thing on the real data

Every reporter in `trade_summary.csv` is in `countries.xlsx`, so all four joins give the same rows. To see the difference, pretend the reference table is incomplete: drop Australia from the lookup. This is the common real-world case, a lookup that does not cover every code in your data.

````python
incomplete = countries[countries["iso3"] != "AUS"]
print("countries:", len(countries), " incomplete lookup:", len(incomplete))
````

### Left join: Australia stays, with a NaN income group

All 280 trade rows survive. The 20 Australia rows have nothing to match, so `income_group` is NaN for them. You can see the problem and decide what to do.

````python
left = trade_dataframe.merge(
    incomplete[["iso3", "income_group"]],
    left_on="reporter_iso3", right_on="iso3", how="left")

print("trade rows:", len(trade_dataframe), " after left join:", len(left))
print("rows with no income group:", left["income_group"].isna().sum())
left.loc[left["income_group"].isna(),
         ["reporter_iso3", "reporter", "year", "product_group", "income_group"]].head()
````

### Inner join: Australia disappears

Leave out `how=` and pandas uses `inner`. The 20 Australia rows are dropped without any warning. A group-by on the result would simply be missing Australia.

````python
inner = trade_dataframe.merge(
    incomplete[["iso3", "income_group"]],
    left_on="reporter_iso3", right_on="iso3")   # how="inner" is the default

print("trade rows:", len(trade_dataframe), " after inner join:", len(inner))
print("Australia in result:", (inner["reporter"] == "Australia").any())
````

### Right join: every country, even with no trade rows

Now the gap is on the other side. Join only the **Africa** trade rows to the full `countries` table. All 14 countries come back; the 11 non-African ones have NaN in the trade columns.

````python
africa = trade_dataframe[trade_dataframe["region"] == "Africa"]

right = africa.merge(
    countries[["iso3", "country_name", "income_group"]],
    left_on="reporter_iso3", right_on="iso3", how="right")

# count() ignores NaN, so this shows how many trade rows each country matched
right.groupby("country_name")["exports_usd_m"].count()
````

### Outer join with `indicator=True`

Both gaps at once: Africa trade rows joined to the incomplete lookup. `indicator=True` adds a `_merge` column saying where each row came from: `both`, `left_only` or `right_only`. This is the quickest way to audit a join.

````python
outer = africa.merge(
    incomplete[["iso3", "country_name", "income_group"]],
    left_on="reporter_iso3", right_on="iso3",
    how="outer", indicator=True)

print(outer["_merge"].value_counts(), end="\n\n")

# which countries sit on only one side?
(outer.loc[outer["_merge"] != "both", ["reporter_iso3", "iso3", "_merge"]]
      .drop_duplicates())
````

### Cross join

No key at all: every trade row paired with every country. 280 x 14 = 3,920 rows. Rarely what you want, but it is there.

````python
len(trade_dataframe.merge(countries, how="cross"))
````

# Making Requests to APIs

````python
import requests
import pandas as pd

response = requests.get("https://api.worldbank.org/v2/country/KEN/indicator/NE.EXP.GNFS.ZS?format=json&date=2015:2023")
data = response.json()
print(data)

# retrieve the second item in the data array
ratio_list = data[1]


# dataframe direct from the API with nested JSON structures in it
simple_dataframe_from_api = pd.DataFrame(ratio_list)
# pull out the nested structures into new columns
dataframe = pd.json_normalize(ratio_list)

print(dataframe.head(3))

# ----------------------------------------------------------------------
# Calling an API that needs a password (an "API key")
# ----------------------------------------------------------------------
# Many APIs give you a single secret key. You send it either as a
# query parameter or, more commonly, as a header. postman-echo.com just
# echoes back whatever it receives, so we can see what was sent.
API_KEY = "demo-key-123"   # in real code: read this from an env variable, never hard-code it

# Option A: key as a query parameter  ->  ...?api_key=demo-key-123
response = requests.get("https://postman-echo.com/get", params={"api_key": API_KEY})
print(response.status_code, response.json()["args"])

# Option B: key in a header (preferred, it does not end up in logs / browser history)
response = requests.get("https://postman-echo.com/get", headers={"X-API-Key": API_KEY})
print(response.status_code, response.json()["headers"]["x-api-key"])

# ----------------------------------------------------------------------
# Calling an API with a username AND password (HTTP Basic authentication)
# ----------------------------------------------------------------------
# requests builds the "Authorization: Basic <base64 of user:password>" header for us.
# This postman-echo endpoint only accepts username "postman" with password "password".
response = requests.get("https://postman-echo.com/basic-auth",
                        auth=("postman", "password"))
print(response.status_code, response.json())      # 200 {'authenticated': True}

# Wrong password -> 401 Unauthorized
response = requests.get("https://postman-echo.com/basic-auth",
                        auth=("postman", "wrong-password"))
print(response.status_code, response.reason)      # 401 Unauthorized

# Always check the status before trusting the body
if response.ok:
    print(response.json())
else:
    print("Request failed:", response.status_code, response.reason)
````
