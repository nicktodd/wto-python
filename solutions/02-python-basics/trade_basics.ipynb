# Part D: Functions

## D1. Your first function
Wrap the cleaning logic in a function so it can be reused. Complete `clean_name(name)` so that it **returns** the cleaned name.

````python
def clean_name(name):
    """Remove surrounding spaces and use title case."""
    return name.strip().title()

print(clean_name("  kenya "))
````

````python
# check
assert clean_name(" kenya") == "Kenya"
assert clean_name("SOUTH AFRICA ") == "South Africa"
print("D1 OK")
````

## D2. Use the function in a loop
Rewrite C4 using your function: build `clean_names2` by calling `clean_name()` inside the loop.

````python
# check
assert clean_names2 == clean_names
print("D2 OK")
````

## D3. A function with a decision
Write `to_number(value)`. If `value` is text (`isinstance(value, str)` is `True`), remove commas and convert it with `float()`. Otherwise just return `float(value)`.

````python
def to_number(value):
    """Convert a number or text such as '14,600' to a float."""
    if isinstance(value, str):
        return float(value.replace(",", ""))
    return float(value)

print(to_number("14,600"), to_number(7412.5))
````

````python
# check
assert to_number("14,600") == 14600.0
assert to_number(7412.5) == 7412.5
print("D3 OK")
````

## D4. Classify a trade balance
Write `classify_balance(exports, imports)` that returns `"Surplus"`, `"Deficit"` or `"Balanced"` using `if`, `elif` and `else`.

````python
def classify_balance(exports, imports):
    """Describe the trade balance."""
    if exports > imports:
        return "Surplus"
    elif exports < imports:
        return "Deficit"
    else:
        return "Balanced"

print(classify_balance(10, 5))
````

````python
# check
assert classify_balance(10, 5) == "Surplus"
assert classify_balance(5, 10) == "Deficit"
assert classify_balance(5, 5) == "Balanced"
print("D4 OK")
````

# Part F: Putting it together

## F1. A list of records
Now we have eight records: a list of dictionaries, like a small table. Run the cell, then complete the loop to print each reporter **as it appears** in the data.

````python
trade = [
    {"reporter": " kenya", "year": 2022, "exports": 7950.0, "imports": 20100.0},
    {"reporter": "Kenya", "year": 2023, "exports": 7412.5, "imports": 21800.0},
    {"reporter": "NIGERIA", "year": 2022, "exports": 63300.0, "imports": 51900.0},
    {"reporter": "Nigeria ", "year": 2023, "exports": 55100.0, "imports": 49700.0},
    {"reporter": "south africa", "year": 2022, "exports": 123600.0, "imports": 113200.0},
    {"reporter": "South Africa", "year": 2023, "exports": 110200.0, "imports": 108900.0},
    {"reporter": "Ghana", "year": 2022, "exports": 17300.0, "imports": "14,600"},
    {"reporter": "ghana", "year": 2023, "exports": 16800.0, "imports": "15,100"},
]
for r in trade:
    print(repr(r["reporter"]))
````

## F2. Clean every record
Build `clean_trade`: for each record, create a **new** dictionary with the cleaned reporter, the year, and exports and imports converted with `to_number()`, and append it.

````python
clean_trade = []
for r in trade:
    new = {
        "reporter": clean_name(r["reporter"]),
        "year": r["year"],
        "exports": to_number(r["exports"]),
        "imports": to_number(r["imports"]),
    }
    clean_trade.append(new)
print(clean_trade[6])
````

````python
# check
assert len(clean_trade) == 8
assert clean_trade[0]["reporter"] == "Kenya"
assert clean_trade[6]["imports"] == 14600.0
print("F2 OK")
````

## F3. A report line per record
Loop over `clean_trade` and print one line per record, e.g. `Kenya 2023: Deficit of 14,387.5`. Use `classify_balance()`, `abs()` and an f-string with `:,.1f`.

````python
for r in clean_trade:
    balance = r["exports"] - r["imports"]
    status = classify_balance(r["exports"], r["imports"])
    print(f"{r['reporter']} {r['year']}: {status} of {abs(balance):,.1f}")
````

## F4. Total exports for one year
Loop over `clean_trade` and add up exports **only** for records where the year is 2023 (use `if` inside the loop).

````python
total_2023 = 0
for r in clean_trade:
    if r["year"] == 2023:
        total_2023 = total_2023 + r["exports"]
print(total_2023)
````

````python
# check
assert total_2023 == 189512.5
print("F4 OK")
````

## Stretch: growth by country
Build a dictionary `exports_by_country` of the form `{"Kenya": {2022: 7950.0, 2023: 7412.5}, ...}`, then print each country's 2022 to 2023 export growth to one decimal place. Hint: `exports_by_country.setdefault(name, {})` returns the inner dictionary, creating it if needed.

````python
exports_by_country = {}
for r in clean_trade:
    exports_by_country.setdefault(r["reporter"], {})[r["year"]] = r["exports"]

for country, years in exports_by_country.items():
    g = (years[2023] - years[2022]) / years[2022] * 100
    print(f"{country}: {g:.1f}%")
````
