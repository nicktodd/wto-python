## Step 6: Total trade by year
`groupby("year")` groups the rows for each year, and `sum()` adds up exports and imports within each group. This is what a pivot table does in Excel.

```python
by_year = df.groupby("year")[["exports_usd_m", "imports_usd_m"]].sum().round(0)
by_year
```

**Question:** What happened to trade in 2020, and why might that be? What happened next?

*Your answer:* Trade fell in 2020 (COVID-19 disruption to demand and supply chains), rebounded strongly in 2021 and peaked in 2022, partly reflecting higher commodity prices.

## Step 8: Exports by region in 2023
This cell filters to 2023, groups by region and sorts the result from largest to smallest.

```python
y2023 = df[df["year"] == 2023]
y2023.groupby("region")["exports_usd_m"].sum().sort_values(ascending=False).round(0)
```

**Question:** Which region exported the most in 2023? Is that a fair comparison between regions? (Think about how many economies are in each region in this dataset.)

*Your answer:* Asia-Pacific. It is not a fair comparison of regions in general: the dataset holds only 14 economies, and Asia-Pacific includes China, Japan, India, Australia and Viet Nam.
