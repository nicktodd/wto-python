## Load a dataset
`pandas` reads a CSV file into a table called a **DataFrame**.

```python
import pandas as pd

df = pd.read_csv(
    "../../data/trade_summary.csv")
cols = ["reporter", "year", "exports_usd_m"]
df[cols].head()
```

```python
print(df.shape)
print(list(df.columns))
```

## One statement, one answer
Total exports for each year: like a pivot table.

```python
by_year = (
    df.groupby("year")["exports_usd_m"]
    .sum()
    .round(0)
)
by_year
```
