## Load a dataset
`pandas` reads a CSV file into a table called a **DataFrame**.

```python
import pandas as pd

df = pd.read_csv(
    "../../data/trade_summary.csv")
df.head()
```

```python
print(df.shape)
print(list(df.columns))
```

## One line, one answer
Total exports by region and year (a pivot table in one line).

```python
totals = (
    df.groupby(["region", "year"])
    ["exports_usd_m"]
    .sum()
    .unstack()
    .round(0)
)
totals
```
