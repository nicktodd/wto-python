## Variables and data types

```python
country = "Kenya"
year = 2023
exports_usd_m = 7412.5
is_wto_member = True
print(type(country), type(year))
print(type(exports_usd_m))
print(type(is_wto_member))
```

## Strings

```python
name = "  viet nam "
clean = name.strip().title()
print(clean)
print(f"{clean}: {exports_usd_m:,.1f} USD m")
print("Nam" in clean)
print(clean.upper(), len(clean))
```

## Numbers and conversion

```python
exports_2022 = 7_950.0
exports_2023 = 7_412.5
change = exports_2023 - exports_2022
growth = change / exports_2022 * 100
print(f"Growth: {growth:.1f}%")

raw = "1,234.5"
value = float(raw.replace(",", ""))
print(value * 2)
```

## Lists

```python
reporters = ["Kenya", "Nigeria", "Ghana"]
reporters.append("Egypt")
print(reporters[0], reporters[-1])
print(len(reporters))

exports_bn = [7.4, 55.1, 17.3, 42.0]
print(sum(exports_bn), max(exports_bn))
print(sorted(exports_bn, reverse=True))
```

## Dictionaries

```python
record = {
    "reporter": "Kenya",
    "year": 2023,
    "exports_usd_m": 7412.5,
}
print(record["reporter"])
record["imports_usd_m"] = 21800.0

for key, value in record.items():
    print(key, "->", value)
```

## Conditional logic

```python
exports = 7412.5
imports = 21800.0
balance = exports - imports

if balance > 0:
    print("Surplus")
elif balance < 0:
    print("Deficit")
else:
    print("Balanced")
```

## Loops

```python
records = [
    {"country": "Kenya", "value": 7.4},
    {"country": "Nigeria", "value": 55.1},
    {"country": "Ghana", "value": 17.3},
]
total = 0
for r in records:
    total += r["value"]
    print(r["country"], r["value"])
print("Total:", round(total, 1))
```

```python
# A list comprehension: filter in one line
big = [r["country"] for r in records
       if r["value"] > 10]
print(big)
```

## Functions

```python
def growth_rate(old, new):
    """Percentage change from old to new."""
    if old == 0:
        return None
    return (new - old) / old * 100

print(growth_rate(7950.0, 7412.5))
print(round(growth_rate(100, 110), 1))
print(growth_rate(0, 5))
```
