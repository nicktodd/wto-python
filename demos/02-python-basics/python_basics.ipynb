## Accumulating a total

```python
exports_bn = [7.4, 55.1, 17.3]
total = 0
for value in exports_bn:
    total = total + value
    print("running total:", round(total, 1))
print("Total:", round(total, 1))
```

## Loops with conditions

## List comprehensions

```python
messy = [" kenya", "NIGERIA", "ghana "]
clean = [n.strip().title() for n in messy]
print(clean)

values = [7.4, 55.1, 17.3]
big = [v for v in values if v > 10]
print(big)
```

## Defining a function

```python
def clean_name(name):
    """Remove spaces; use title case."""
    return name.strip().title()

print(clean_name("  kenya "))
print(clean_name("SOUTH AFRICA"))
```

## Parameters and return values

```python
def growth_rate(old, new):
    """Percentage change from old to new."""
    if old == 0:
        return None
    return (new - old) / old * 100

g = growth_rate(7950.0, 7412.5)
print(round(g, 1))
print(growth_rate(0, 5))
```

## Default and keyword arguments

```python
def fmt(value, unit="USD m", dp=1):
    return f"{value:,.{dp}f} {unit}"

print(fmt(7412.5))
print(fmt(7.4125, unit="USD bn"))
print(fmt(7412.5, dp=0))
```

## return versus print

```python
def show_total(values):
    print(sum(values))

def get_total(values):
    return sum(values)

a = show_total([1, 2, 3])
b = get_total([1, 2, 3])
print("a =", a, "| b =", b)
```

## Putting it together

```python
def clean_record(r):
    name = clean_name(r["reporter"])
    return {"reporter": name,
            "exports": float(r["exports"])}

raw = [
    {"reporter": " kenya", "exports": "7.4"},
    {"reporter": "GHANA", "exports": "17.3"},
]
clean = [clean_record(r) for r in raw]
for r in clean:
    print(r)
```
